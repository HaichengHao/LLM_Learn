```python
import pandas as pd
```

```python
advertising=pd.read_csv('./data_collections/advertising.csv')
advertising.drop(advertising.columns[0], axis=1, inplace=True) #axis指定为1去掉第一列
advertising.dropna(inplace=True) #去除缺失值
X=advertising.drop('Sales',axis=1) #去除标签列Sales
y=advertising['Sales']
```

```python
print(X.corrwith(y,method='pearson'))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

#计算皮尔逊相关系数矩阵
corr_matrix=advertising.corr(method='pearson')
sns.heatmap(corr_matrix,annot=True,cmap='coolwarm',fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.show()
```
